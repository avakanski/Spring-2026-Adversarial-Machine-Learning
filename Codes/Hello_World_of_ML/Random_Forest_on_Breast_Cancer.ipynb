# Random Forest Classification of the Breast Cancer Dataset

This notebook demonstrates how to use a Random Forest classifier to predict breast cancer diagnosis. The dataset contains 569 samples with 30 features computed from digitized images of breast mass cells.

Target Variable:
- 0 = Malignant (cancerous)
- 1 = Benign (non-cancerous)

## Import Modules

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report, confusion_matrix
import pandas as pd
```

## Load the Breast Cancer Dataset

```python
# Load the dataset from sklearn datasets
bc = load_breast_cancer()
```

```python
# Convert to Pandas data frame and display the first 10 rows
df = pd.DataFrame(data=bc.data, columns=bc.feature_names)
df["target"] = bc.target
df.head(10)
```

```python
# Print data shape (569 samples, 31 features)
df.shape
```

```python
# Show the features and other basic information
df.info()
```

```python
# Check the distribution of target values
df['target'].value_counts()
# There are 357 patients with malignant tumors (label 1), and 212 patients with benign tumors (label 0)
```

## Prepare Training and Testing Datasets

```python
# Define features (X) and targets (y) variables
X = df.drop('target', axis=1)
y = df['target']
```

```python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=123, stratify=y)
```

```python
# Print the shapes of data and label sets
print('Training data features', X_train.shape)
print('Training labels', y_train.shape)
print('Testing data features', X_test.shape)
print('Testing labels', y_test.shape)
```

## Train a Random Forest Model

```python
# Create an instance of a Random Forest model
model = RandomForestClassifier(random_state=42)
```

```python
# Fit the model to the training dataset
model.fit(X_train, y_train)
```

## Evaluate the Model

```python
# Predict on test dataset
model_predictions = model.predict(X_test)
```

```python
# Calculate and print the accuracy
accuracy = accuracy_score(y_test, model_predictions)
print(f"Accuracy on test dataset: {accuracy*100:5.2f}%")
```

```python
# Print confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, model_predictions))
```

The confusion matrix shows:
- True Negatives (top-left): Correctly predicted benign cases
- False Positives (top-right): Benign cases incorrectly predicted as malignant
- False Negatives (bottom-left): Malignant cases incorrectly predicted as benign
- True Positives (bottom-right): Correctly predicted malignant cases

```python
# Print classification report
print("Classification report:")
print(classification_report(y_test, model_predictions))
```

The report provides detailed metrics:
- Precision: Of all predicted positives, how many were correct?
- Recall: Of all actual positives, how many were detected?
- F1-Score: Harmonic mean of precision and recall.
