# Logistic Regression Classifiction on the Iris Dataset

This notebook demonstrates how to use Logistic Regression to classify iris flowers into three species. The dataset contains 150 samples with 4 features measuring the dimensions of iris flowers.

Dataset Information:
- 150 samples total (50 per species)
- 4 features: sepal length, sepal width, petal length, petal width
- 3 target classes: Setosa, Versicolor, Virginica

Target Classes:
- 0 = Iris Setosa
- 1 = Iris Versicolor
- 2 = Iris Virginica

## Import Modules

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report, confusion_matrix
```

## Load the Iris Dataset

```python
# Load the dataset from sklearn datasets
iris = load_iris()
```

```python
# Print data shape (150 samples, 4 features)
iris.data.shape
```

```python
# Print the names of the 4 features
print(iris.feature_names)
```

```python
# Print the categories of Iris flowers
# 0 is Iris Setosa, 1 is Iris Versicolor, 2 is Iris Virginica
print(iris.target)
```

## Prepare Training and Testing Datasets

```python
# Define features (X) and targets (y) variables
X, y = iris.data, iris.target
```

```python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=123, test_size=0.25, stratify=y)
```

```python
# Print the shapes of data and label sets
print('Training data features', X_train.shape)
print('Training labels', y_train.shape)
print('Testing data features', X_test.shape)
print('Testing labels', y_test.shape)
```

## Train a Logistic Regression Model

```python
# Create an instance of a Logistic Regression model
model = LogisticRegression(max_iter=1000)
```

```python
# Fit the model to the training dataset
model.fit(X_train, y_train)
```

## Evaluate the Model

```python
# Predict on test dataset
logreg_predictions = model.predict(X_test)
```

```python
# Calculate and print the accuracy
accuracy = accuracy_score(y_test, logreg_predictions)
print(f"Accuracy on test dataset: {accuracy*100:5.2f}%")
```

```python
# Print confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, logreg_predictions))
```

The confusion matrix shows how well the model predicted each species:
- Rows: Actual species
- Columns: Predicted species
- Diagonal values: Correct predictions
- Off-diagonal values: Misclassifications

```python
# Print classification report
print("Classification report:")
print(classification_report(y_test, logreg_predictions, target_names=iris.target_names))
```

The report provides detailed metrics:
- Precision: Of all predicted positives, how many were correct?
- Recall: Of all actual positives, how many were detected?
- F1-Score: Harmonic mean of precision and recall.
